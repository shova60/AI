```python
name = "Shova"
age = 21
department = "CSE"

print("Name:", name)
print("Age:", age)
print("Department:", department)

a = 30
b = 10

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
a = 20
b = 15

if a > b:
    print("A is largest")
elif b > a:
    print("B is largest")
else:
    print("Both are equal")
```

```python
print("For Loop:")

for i in range(1, 8):
    print(i)


# While Loop

print("While Loop:")

i = 1

while i <= 8:
    print(i)
    i = i + 1
```

```python
def add(a, b):
    return a + b


def multiply(a, b):
    return a * b


result1 = add(70, 20)
result2 = multiply(5, 9)

print("Addition:", result1)
print("Multiplication:", result2)
```
