```python
import matplotlib.pyplot as plt

x = ["Python", "AI", "Math", "English"]
y = [80, 90, 70, 85]

plt.bar(x, y)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Subject Marks")

plt.show()
```

```python

import matplotlib.pyplot as plt

subjects = ["Python", "AI", "Math", "English"]
marks = [88, 90, 85, 85]

plt.pie(marks, labels=subjects, autopct='%1.1f%%')

plt.title("Subject Marks Distribution")

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [6, 7, 8, 9, 10]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Simple Line Graph")

plt.show()
```

```python

import matplotlib.pyplot as plt

x = [1, 3, 6, 9, 12]
y = [5, 10, 8, 15, 12]

plt.scatter(x, y)

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Scatter Plot")

plt.show()
```

```python

import matplotlib.pyplot as plt

x = [6, 7, 8, 9, 10]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, marker='o')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Simple Line Graph")

plt.savefig("my_graph.png", dpi=300)
plt.show()
```
